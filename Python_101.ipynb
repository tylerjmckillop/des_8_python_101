## 2) Packages and Libraries

This is how you can obtain your datetime, tables, dataframes, and more. There's over 500,000 python packages and libraries out there in the wild. But what are they?

**Library**

- Collection of useful codes you can reuse

  - Tools
  - For example NumPy (Number python) is a library that allows you to create arrays for example (Lists but of all numbers where you can execute advanced mathematical operations)

**Packages**

- Collection of related python files (modules) grouped together in a folder

Libraries and packages can be public or native

Native don't required installation and just need to be imported in your code such as

- math (mathematical functions)
- random (generate random numbers)
- datetime (datetime data types, functions etc.)

How does one install a public package?

## 3) Loops

There are two main types of loops:

**For Loops**

Syntax:

```python
for x in y:
    code block referencing x
```


- Keep doing a task until you have exhausted the list i have provided
- Automate repetitive tasks
- Makes code shorter and avoid repeating yourself
- **Indentation is important!!!!!!**
- Syntax has 3 key features as `for x in y:
  - **for** : tells python you are starting a foor loop
  - **x**: what you call this parameter is up to you, it's what you would call one item in the list **y**
  - **y**: this is the collection of items you want to iterate thought, or loop through
- The code block referencing x needs to be indented, all lines.

For example:


````python
#Creating a list
shopping_list = ["milk", "bread", "eggs"] 

#Looping through list
for item in shopping_list:
    #each time print the item to buy
    print(f'Remember to buy {item}')
````

````python
#Creating a list
numbers = [4,8,15,16, 23, 42]
#Define variable before the for loop!
total =0

#loop through the numbers
for num in numbers:
    #sum up our numbers, += operator is an easier way to write total = total + number here
    total+= num

print(total)
````

````python
#You can also loop through a range of numbers: 
#range(x,y) creates a list of numbers to loop through, the second number is NOT INCLUDED
for number in range(1,6):
    print(number*number)
````

````python
#Loop through a list of dictionaries 
#Creating a list of dictionaries
user_dictionary= [
    {"name": "Jenny", "active": True},
    {"name": "James", "active": False},
    {"name": "Jules", "active": True}
]

#Loop through the list
for user in user_dictionary:
    #If statement to only print active users, if user["active"] is equivalent to if user["active"] = True
    if user["active"]:
        print(f"{user['name']} is active")

````

**While Loops**

Syntax:

```python
while condition:
    code block
```


- Keep doing a task until the condition is FALSE
- Automate repetitive tasks
- Makes code shorter and avoid repeating yourself
- **Indentation is important!!!!!!**
- Syntax has 2 key features as while condition:
  - **while** : tells python you are starting a while loop
  - **condition**: set your condition that will break the loop, as long as it's true the loop will repeat, as soon as this becomes false, the code will stop iterating
- The whole code block needs to be indented or it will be considered out of the loop
- Within a while loop you can add a break or continue statements, break will lead you out of the loop, continue will skip an iteration and keep the while loop iterating through the next run, you can use these powerfully along if statements.

For example:


````python
#Create a count variable
count = 0

#Looping until count is over 5
while count <= 5: 
    count+=1
    if count == 3:
        #Use continue statement, notice how 3 isn't printed? 
        continue
    print(count)
    
````

## 4) Functions

A function is a reusable block of code that performs a repeated task. You define the function once and then you can use or call it as many times as you want and even across different python scripts.

Syntax:

```python
def function_name (parameter 1, parameter 2....):
    code block reference parameter 1
    code block reference parameter 2
    return result/output of the function
```


- Makes code cleaner and easier to read
- Avoids repetition
- Helps with testing and debugging
- Makes your code modular and reusable
- Syntax includes 4 key features
  - **def**: begins the function
  - **function_name**: name of your function, make it easy to understand, you will need to use it whenever you want to run the function
  - **parameters**: optional inputs that your functions uses (you can make them necessary to run the functions or optional with a default value in case there is no value inputed)
  - **return**: optional modified that returns an output of your function

For Example:


````python
#Addition Function

def add(a,b):
    return a+b
print(add(4,5))
````

````python
#Currency conversion

#set function with default rate value)
def euros_to_pound(eur, rate=0.87):
    return eur*rate

#Print with default rate
print(euros_to_pound(100))

#Print with set rate
print(euros_to_pound(100,0.75))
````

````python
#Categorisation

def categorise_spend(amount, low_limit=50, high_limit = 200):
    if amount < low_limit:
        return 'Low Spending'
    elif low_limit <= amount < high_limit:
        return 'Medium Spending'
    else:
        return 'High Spending'

print(categorise_spend(150, low_limit = 200, high_limit = 400))
````

## 5) Try/Except Blocks

Try/except blocks are used to handle errors (exceptions) gracefully so your program doesn’t crash when something unexpected happens. You put the code that might fail inside the try block, and if an error occurs, Python jumps to the except block where you can decide how to handle it — such as logging a message, using a fallback value, or safely continuing the program.

For example:

````python
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(f"Result is {result}")
except Exception as e:
    print(f"Something went wrong: {e}")
````

## 6) Logging

Logging helps us retain an audit of how our scripts run to:

- Debug failures (API failures, rate limits, etc.)
- Monitor performance of pipeline
- See what happened, when

`logging` is a part of Python's standard library. (no need for pip install)
[https://docs.python.org/3/library/logging.html](https://docs.python.org/3/library/logging.html)

##### When to use logging

You can access logging functionality by creating a logger via `logger = getLogger()`, and then calling the logger’s `debug()`, `info()`, `warning()`, `error()` and `critical()` methods. To determine when to use logging, and to see which logger methods to use when, see the table below. It states, for each of a set of common tasks, the best tool to use for that task.

| Task you want to perform | The best tool for the task |
|---------------------------|----------------------------|
| Display console output for ordinary usage of a command line script or program | `print()` |
| Report events that occur during normal operation of a program (e.g. for status monitoring or fault investigation) | A logger's `info()` (or `debug()` method for very detailed output for diagnostic purposes) |
| Issue a warning regarding a particular runtime event | `warnings.warn()` in library code if the issue is avoidable and the client application should be modified to eliminate the warning<br><br>A logger's `warning()` method if there is nothing the client application can do about the situation, but the event should still be noted |
| Report an error regarding a particular runtime event | Raise an exception |
| Report suppression of an error without raising an exception (e.g. error handler in a long-running server process) | A logger's `error()`, `exception()` or `critical()` method as appropriate for the specific error and application domain |

The logger methods are named after the level or severity of the events they are used to track. The standard levels and their applicability are described below (in increasing order of severity):

| Level | When it's used |
|-------|----------------|
| `DEBUG` | Detailed information, typically of interest only when diagnosing problems. |
| `INFO` | Confirmation that things are working as expected. |
| `WARNING` | An indication that something unexpected happened, or indicative of some problem in the near future (e.g. 'disk space low'). The software is still working as expected. |
| `ERROR` | Due to a more serious problem, the software has not been able to perform some function. |
| `CRITICAL` | A serious error, indicating that the program itself may be unable to continue running. |

##### Basic Config for logs

Default settings and key arguments 
```
logging.basicConfig(
    filename = "filename",
    filemode = "a", # Append new logs to file by default, can be "w" to overwrite the file
    format = "levelname:name:message", # format for content output in log file
    level=logging.WARNING, # minimum serverity for log to be recorded in a log file
)
```

##### Assign Logger
logger = logging.getLogger()

##### Set Logging Messages
Recall the Log Level serverity hierarchy: DEBUG < INFO < WARNING < ERROR < CRITICAL

Applied through these methods
- logger.debug("")
- logger.info("")  
- logger.warning("")       
- logger.error("") 
- logger.critical("") 

Setting `level=logging.INFO` will only capture message of status INFO and higher

##### Simple Logging Example
Make a folder called "logs" first. Run the script. Then take a look at the file it creates. 

````python
import logging
from datetime import datetime

# Set Log filepath location
log_filename = f"logs/logging_example_{datetime.now().strftime('%Y%m%d_%H%M%S')}.log"

# Configure logs to retrieve INFO messages and higher
logging.basicConfig(
    level=logging.INFO, 
    format='%(asctime)s - %(levelname)s - %(message)s',
    filename=log_filename
)

logger = logging.getLogger()

#Add the following to your script when you want logging to occur:
logger.debug("This is a debug message")    
logger.info("System working")            
logger.warning("Something unexpected")        
logger.error("An error occurred")             
logger.critical("Critical system error")
````
